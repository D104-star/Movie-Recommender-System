```python
#'[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
#Our target is to convert the above form to the below form
#['Action','Adventure','Fantasy','Science Fiction']
#'[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
#this is actually a string of lists which has to be first converted to a list to perform the above wanted conversion
#for that we have a module ast,in which there is a function literal_eval() which exatly performs this task and gives us a list
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
)
```

```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
#Performing the same steps on keywords as on the genres
movies.iloc[0].keywords
```

```python
import ast
ast.literal_eval('[{"id": 1463, "name": "culture clash"}, {"id": 2964, "name": "future"}, {"id": 3386, "name": "space war"}, {"id": 3388, "name": "space colony"}, {"id": 3679, "name": "society"}, {"id": 3801, "name": "space travel"}, {"id": 9685, "name": "futuristic"}, {"id": 9840, "name": "romance"}, {"id": 9882, "name": "space"}, {"id": 9951, "name": "alien"}, {"id": 10148, "name": "tribe"}, {"id": 10158, "name": "alien planet"}, {"id": 10987, "name": "cgi"}, {"id": 11399, "name": "marine"}, {"id": 13065, "name": "soldier"}, {"id": 14643, "name": "battle"}, {"id": 14720, "name": "love affair"}, {"id": 165431, "name": "anti war"}, {"id": 193554, "name": "power relations"}, {"id": 206690, "name": "mind and soul"}, {"id": 209714, "name": "3d"}]'
                )
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
#for cast, character is the name of the actor in the movie,but we just want the actual name of the actor and we just want the names of top 3 actors
def convert3(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
     if counter !=3:
        L.append(i['name'])
        counter+=1
     else:
        break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
#Now in case of the crew out of so many dictionaries we want only that dictionary which has the value of job as director
def fetch_director(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
         L.append(i['name'])
         break
    return L
```

```python
movies['crew'].apply(fetch_director)
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
#Now we also have to convert the column of Overview into a list so that it can be concatinated with the rest of the lists
movies['overview'][0]
```

```python
movies['overview'].apply(lambda x:x.split())
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
#Now we have extracted the useful words from all the columns
#Now the next problem that arises is that we have to remove the space between the words
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
#Now we have to create a new column of tags where we have to concatenate the columns of overview,genres,keywords,cast,crew
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
#Now we don't want the rest of the columns, os we'll create a new dataset containing only the required columns
new_df=movies[['id','title','tags']]
```

```python
#Converting list in tags to a string
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
     y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'][0]
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
#It's generally recommended to convert everything into the lowercase
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
#Now we have to create a function that if a movie is provide it will further return the names of 5 similar movies
def recommend(movie):
    movie_index=new_df[new_df['title']==movie].index[0]
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
    return    
```
